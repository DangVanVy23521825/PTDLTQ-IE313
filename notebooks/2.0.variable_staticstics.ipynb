# Variable Statistics Analysis

Phân tích thống kê mô tả cho các biến trong dataset YouTube videos sau khi đã làm sạch dữ liệu.

## 1. Load Data & Variable Classification

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
# Load cleaned data
df = pd.read_csv('data/youtube_clean.csv')
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"\nColumns:\n{list(df.columns)}")
df.head()
```

```python
# Phân loại các biến
numerical_cols = ['duration', 'bitrate', 'bitrate(video)', 'height', 'width',
                  'frame rate', 'frame rate(est.)', 'views', 'likes', 'comments']

categorical_cols = ['title', 'description', 'hashtags', 'codec', 'category']

print("NUMERICAL VARIABLES")
print(f"Total: {len(numerical_cols)} columns")
print(f"Columns: {numerical_cols}\n")
print("CATEGORICAL VARIABLES")
print(f"Total: {len(categorical_cols)} columns")
print(f"Columns: {categorical_cols}")
```

## 2. Descriptive Statistics

### 2.1 Numerical Variables

```python
# Thống kê mô tả cho biến numerical
desc = df[numerical_cols].describe()
desc.loc['median'] = df[numerical_cols].median()
desc.loc['skewness'] = df[numerical_cols].skew()
desc.loc['kurtosis'] = df[numerical_cols].kurtosis()

desc = desc.round(2)
desc
```

### 2.2 Categorical Variables

```python
# Thống kê mô tả cho biến categorical
cat_stats = pd.DataFrame({
    'Column': categorical_cols,
    'Unique Values': [df[col].nunique() for col in categorical_cols],
    'Most Frequent': [df[col].mode()[0] if len(df[col].mode()) > 0 else None for col in categorical_cols],
    'Frequency': [df[col].value_counts().iloc[0] if len(df[col]) > 0 else 0 for col in categorical_cols],
    'Percentage (%)': [round(df[col].value_counts().iloc[0] / len(df) * 100, 2) if len(df[col]) > 0 else 0 for col in categorical_cols]
})

cat_stats.set_index('Column', inplace=True)
cat_stats
```
