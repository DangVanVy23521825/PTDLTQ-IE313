Check các cột nghi ngờ cho giá trị null ẩn không thể kiểm tra bằng isnull()

```python
def check_hidden_missing(df):
    suspected = ['?', '.', 'N/A', 'NA', 'NaN', 'nan', 'NULL', 'Null', 'None', ' ', '', '0']
    result = {}

    for col in df.columns:
        mask = df[col].astype(str).isin(suspected) | df[col].astype(str).str.strip().eq('')
        result[col] = mask.sum()

    result_df = pd.DataFrame.from_dict(result, orient='index', columns=['Suspected Missing'])
    result_df['Percentage (%)'] = 100 * result_df['Suspected Missing'] / len(df)

    return result_df[result_df['Suspected Missing'] > 0].sort_values('Percentage (%)', ascending=False)

hidden_missing_table = check_hidden_missing(df)
hidden_missing_table
```

```python
df.loc[df['bitrate(video)'] == 0, ['duration', 'width', 'height', 'views']].head(10)
```

```python
df.loc[df['bitrate'] == 0, 'bitrate'] = np.nan
df.loc[df['bitrate(video)'] == 0, 'bitrate(video)'] = np.nan
```

```python
suspected_missing = ['?', '.', 'N/A', 'NA', 'NaN', 'nan', 'NULL', 'Null', 'None', ' ', '']

# Chuẩn hóa các giá trị nghi ngờ thành NaN
df['description'].replace(suspected_missing, np.nan, inplace=True)
df['hashtags'].replace(suspected_missing, np.nan, inplace=True)

# Điền lại bằng giá trị thay thế có ý nghĩa
df['description'].fillna('No description', inplace=True)
df['hashtags'].fillna('No hashtags', inplace=True)
```

## Điền khuyết

```python
df_drop = df.copy()
df_fill_median = df.copy()
df_fill_mean = df.copy()
```

```python
df_drop.dropna(subset=['bitrate', 'bitrate(video)'], inplace=True)

df_fill_median['bitrate'].fillna(df_fill_median['bitrate'].median(), inplace=True)
df_fill_median['bitrate(video)'].fillna(df_fill_median['bitrate(video)'].median(), inplace=True)

df_fill_mean['bitrate'].fillna(df_fill_mean['bitrate'].mean(), inplace=True)
df_fill_mean['bitrate(video)'].fillna(df_fill_mean['bitrate(video)'].mean(), inplace=True)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# bitrate
sns.kdeplot(df_drop['bitrate'], label='Dropped NaN', ax=axes[0])
sns.kdeplot(df_fill_median['bitrate'], label='Filled Median', ax=axes[0])
sns.kdeplot(df_fill_mean['bitrate'], label='Filled Mean', ax=axes[0])
axes[0].set_title('Distribution Comparison: bitrate')
axes[0].legend()

# bitrate(video)
sns.kdeplot(df_drop['bitrate(video)'], label='Dropped NaN', ax=axes[1])
sns.kdeplot(df_fill_median['bitrate(video)'], label='Filled Median', ax=axes[1])
sns.kdeplot(df_fill_mean['bitrate(video)'], label='Filled Mean', ax=axes[1])
axes[1].set_title('Distribution Comparison: bitrate(video)')
axes[1].legend()

plt.show()
```

3 đường kde đè lên nhau -> ảnh hưởng shape phân phối là như

```python
compare = pd.DataFrame({
    'Method': ['Dropped', 'Filled Median', 'Filled Mean'],
    'bitrate_mean': [df_drop['bitrate'].mean(), df_fill_median['bitrate'].mean(), df_fill_mean['bitrate'].mean()],
    'bitrate_std': [df_drop['bitrate'].std(), df_fill_median['bitrate'].std(), df_fill_mean['bitrate'].std()],
    'bitrate(video)_mean': [df_drop['bitrate(video)'].mean(), df_fill_median['bitrate(video)'].mean(), df_fill_mean['bitrate(video)'].mean()],
    'bitrate(video)_std': [df_drop['bitrate(video)'].std(), df_fill_median['bitrate(video)'].std(), df_fill_mean['bitrate(video)'].std()]
})
compare
```

bitrate_std và birate(video)_std gần như là như nhau ở cả 3 method -> độ phân tán giống nhau => phương án loại bỏ các bản ghi bị khuyết (Dropped) vì tỷ lệ thiếu nhỏ (<1%) và đảm bảo giữ nguyên phân phối tự nhiên của dữ liệu.
