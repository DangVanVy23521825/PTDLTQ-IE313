```python
import pandas as pd
import numpy as np
youtube_no_outliers_clean_final = pd.read_csv('../data/youtube_no_outliers_clean_final.csv', sep = ',')
youtube_no_outliers_clean_final
```

```python
df = youtube_no_outliers_clean_final
```

Xoá đi cột frame(est.) như đã phân tích

```python
df.drop(columns=['frame rate(est.)', 'frame rate(est.)_log'], inplace=True)
```

Xoá đi cột birate và birate(video) như đã phân tích, giữ lại duy nhất 1 cột birate_log

```python
df.drop(columns=['bitrate', 'bitrate(video)_log', 'bitrate(video)'], inplace=True)
```

```python
df.info()
```

```python
numerical_log_cols = [
    'duration_log', 'height_log', 'width_log', 'frame rate_log',
    'views_log', 'likes_log', 'comments_log',
    'likes_log_win', 'comments_log_win'
]

df_norm = df.copy()

# Áp dụng 3 phương pháp chuẩn hóa
for col in numerical_log_cols:
    col_max = df[col].max()
    col_min = df[col].min()
    col_mean = df[col].mean()
    col_std = df[col].std()

    # Scale (chia tỉ lệ so với max)
    df_norm[f'{col}_scaled'] = df[col] / col_max

    # Min-Max
    df_norm[f'{col}_minmax'] = (df[col] - col_min) / (col_max - col_min)

    # Z-score
    df_norm[f'{col}_zscore'] = (df[col] - col_mean) / col_std
```

```python
df['has_description'] = df['description'].astype(str).str.strip().apply(
    lambda x: 0 if x.lower() in ['none', 'no description', '', ' '] else 1
)

df['has_hashtags'] = df['hashtags'].astype(str).str.strip().apply(
    lambda x: 0 if x.lower() in ['no hashtags', 'none', '', ' '] else 1
)
```

```python
df_encode = df[['codec', 'category', 'has_description', 'has_hashtags']].copy()

df_encode
```

```python
df_encode['is_h264'] = (df['codec'] == 'h264').astype(int)

df_encode = pd.concat([df_encode,
                       pd.get_dummies(df['category'], prefix='cat', drop_first=True)
                      ], axis=1)

df_encode
```

```python
df_model = pd.concat([df_norm, df_encode], axis=1)

df_model
```

```python
df_model.info()
```

```python
cols_to_drop = [
    'duration', 'height', 'width', 'frame rate',
    'codec', 'category', 'description', 'hashtags', 'title'
]
df_model = df_model.drop(columns=cols_to_drop)
```

# Feature engineering

```python
df_model['like_per_view_log'] = df_model['likes_log'] - df_model['views_log']
df_model['comment_per_view_log'] = df_model['comments_log'] - df_model['views_log']

# Engagement rate log: log((likes + comments)/views)
df_model['engagement_rate_log'] = np.log(np.exp(df_model['likes_log']) + np.exp(df_model['comments_log'])) - df_model['views_log']

# Tỷ lệ giữa dimension
df_model['aspect_ratio'] = df_model['width_log'] - df_model['height_log']  # log(width/height)
df_model['log_aspect_ratio'] = df_model['width_log'] - df_model['height_log']

# Kiểm tra các cột mới
df_model[['like_per_view_log', 'comment_per_view_log', 'engagement_rate_log',
          'aspect_ratio', 'log_aspect_ratio']].head()
```
