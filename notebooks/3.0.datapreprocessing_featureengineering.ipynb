Xoá đi cột frame(est.) như đã phân tích

```python
df.drop(columns=['frame rate(est.)', 'frame rate(est.)_log'], inplace=True)
```

Xoá đi cột birate và birate(video) như đã phân tích, giữ lại duy nhất 1 cột birate_log

```python
df.drop(columns=['bitrate', 'bitrate(video)_log', 'bitrate(video)'], inplace=True)
```

```python
df.info()
```

```python
numerical_log_cols = [
    'duration_log', 'height_log', 'width_log', 'frame rate_log',
    'views_log', 'likes_log', 'comments_log',
    'likes_log_win', 'comments_log_win'
]

df_norm = df.copy()

# Áp dụng 3 phương pháp chuẩn hóa
for col in numerical_log_cols:
    col_max = df[col].max()
    col_min = df[col].min()
    col_mean = df[col].mean()
    col_std = df[col].std()

    # Scale (chia tỉ lệ so với max)
    df_norm[f'{col}_scaled'] = df[col] / col_max

    # Min-Max
    df_norm[f'{col}_minmax'] = (df[col] - col_min) / (col_max - col_min)

    # Z-score
    df_norm[f'{col}_zscore'] = (df[col] - col_mean) / col_std
```

```python
df['has_description'] = df['description'].astype(str).str.strip().apply(
    lambda x: 0 if x.lower() in ['none', 'no description', '', ' '] else 1
)

df['has_hashtags'] = df['hashtags'].astype(str).str.strip().apply(
    lambda x: 0 if x.lower() in ['no hashtags', 'none', '', ' '] else 1
)
```

```python
df[['has_description', 'has_hashtags']].value_counts(normalize=True) * 100
```

```python
df_encode = df[['codec', 'category', 'has_description', 'has_hashtags']].copy()

df_encode
```

```python
df_encode['is_h264'] = (df['codec'] == 'h264').astype(int)

df_encode = pd.concat([df_encode,
                       pd.get_dummies(df['category'], prefix='cat', drop_first=True)
                      ], axis=1)

df_encode
```

```python
df_model = pd.concat([df_norm, df_encode], axis=1)

df_model
```

```python
df_model.info()
```

# Feature engineering

```python
# 1. Aspect ratio - HỢP LỆ (chỉ dùng width và height)
df_model['aspect_ratio'] = df_model['width'] / df_model['height']
df_model['log_aspect_ratio'] = df_model['width_log'] - df_model['height_log']  # log(width/height)

# 2. Video complexity metrics - HỢP LỆ
df_model['video_quality_score'] = df_model['height_log'] + df_model['bitrate_log']
df_model['file_size_estimate'] = df_model['duration_log'] + df_model['bitrate_log']

# 3. Frame density - HỢP LỆ
df_model['frame_density'] = df_model['frame rate_log'] + df_model['duration_log']

# 4. Resolution category - HỢP LỆ
df_model['is_hd'] = (df_model['height'] >= 720).astype(int)
df_model['is_full_hd'] = (df_model['height'] >= 1080).astype(int)
df_model['is_4k'] = (df_model['height'] >= 2160).astype(int)

# 5. Duration category - HỢP LỆ
df_model['is_short'] = (df_model['duration'] < 60).astype(int)
df_model['is_medium'] = ((df_model['duration'] >= 60) & (df_model['duration'] < 600)).astype(int)
df_model['is_long'] = (df_model['duration'] >= 600).astype(int)

# 6. Content richness score
df_model['content_score'] = df_model['has_description'] + df_model['has_hashtags']
```

```python
cols_to_drop = [
    'duration', 'height', 'width', 'frame rate',
    'codec', 'category', 'description', 'hashtags', 'title'
]
df_model = df_model.drop(columns=cols_to_drop)
```
