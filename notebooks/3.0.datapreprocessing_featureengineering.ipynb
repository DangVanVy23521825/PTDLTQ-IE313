# Data Preprocessing

```python
from sklearn.preprocessing import StandardScaler

numerical_log_cols = [
    'duration_log', 'height_log', 'width_log', 'frame rate_log', 'bitrate_log',
    'views_log', 'likes_log', 'comments_log',
    'likes_log_win', 'comments_log_win',
]

df_norm = df.copy()

scaler = StandardScaler()

# Fit + transform
df_norm[[f'{col}_scaled' for col in numerical_log_cols]] = scaler.fit_transform(
    df[numerical_log_cols]
)
```

```python
df_norm[[f'{c}_scaled' for c in numerical_log_cols]].describe().T
```

```python
df_norm.shape
```

```python
df['has_description'] = df['description'].astype(str).str.strip().apply(
    lambda x: 0 if x.lower() in ['none', 'no description', '', ' '] else 1
)

df['has_hashtags'] = df['hashtags'].astype(str).str.strip().apply(
    lambda x: 0 if x.lower() in ['no hashtags', 'none', '', ' '] else 1
)
```

```python
df[['has_description', 'has_hashtags']].value_counts(normalize=True) * 100
```

```python
df_encode = df[['codec', 'category', 'has_description', 'has_hashtags']].copy()

df_encode
```

```python
df_encode = pd.concat([df_encode,
                       pd.get_dummies(df['category'], prefix='cat', drop_first=True)
                      ], axis=1)

df_encode
```

```python
df_model = pd.concat([df_norm, df_encode], axis=1)

df_model
```

Xử Lý Height × Width Multicollinearity

```python
df_model['resolution_score_log'] = np.log1p(
    np.sqrt(np.expm1(df_model['height_log']) * np.expm1(df_model['width_log']))
)

df_model['resolution_score_scaled'] = (
    df_model['resolution_score_log'] / df_model['resolution_score_log'].max()
)

cols_to_drop = ['height_log', 'width_log', 'height_log_scaled', 'width_log_scaled', 'resolution_score_log']
df_model = df_model.drop(columns=cols_to_drop, errors='ignore')
```

```python
df_model.info()
```

# Feature engineering

```python
df_model['aspect_ratio'] = df_model['width'] / df_model['height']

df_model['is_hd'] = (df_model['height'] >= 720).astype(int)
df_model['is_full_hd'] = (df_model['height'] >= 1080).astype(int)
df_model['is_4k'] = (df_model['height'] >= 2160).astype(int)

df_model['is_short'] = (df_model['duration'] < 60).astype(int)
df_model['is_medium'] = ((df_model['duration'] >= 60) & (df_model['duration'] < 600)).astype(int)
df_model['is_long'] = (df_model['duration'] >= 600).astype(int)

df_model['content_score'] = df_model['has_description'] + df_model['has_hashtags']

df_model['duration_bitrate_ratio'] = (
    df_model['duration_log_scaled'] / (df_model['bitrate_log_scaled'] + 0.01)
)

df_model['optimal_duration'] = (
    (df_model['duration'] >= 120) & (df_model['duration'] <= 600)
).astype(int)

df_model['content_completeness'] = (
    0.6 * df_model['has_description'] +
    0.4 * df_model['has_hashtags']
)


# Category popularity (từ ANOVA analysis)
# Music, Entertainment, Comedy, Gaming có views cao hơn
popular_cats = ['Music', 'Entertainment', 'Comedy', 'Gaming']
df_model['is_popular_category'] = df_model[[f'cat_{c}' for c in popular_cats]].max(axis=1)

# Frame rate category
# 24fps (cinema), 30fps (standard), 60fps (high)
df_model['is_high_fps'] = (df_model['frame rate'] >= 50).astype(int)
df_model['is_cinema_fps'] = (
    (df_model['frame rate'] >= 23) & (df_model['frame rate'] <= 25)
).astype(int)
```

```python
cols_to_drop = [
    'duration', 'height', 'width', 'frame rate',
    'codec', 'category', 'description', 'hashtags', 'title',
    'duration_log', 'frame rate_log',
    'bitrate_log'
]
df_model = df_model.drop(columns=cols_to_drop)
```

```python
rename_dict = {
    'duration_log_scaled': 'duration_scaled',
    'frame rate_log_scaled': 'framerate_scaled',
    'bitrate_log_scaled': 'bitrate_scaled'
}
df_model = df_model.rename(columns=rename_dict)
```
